```python
import numpy as np
import pandas as pd


p1= {'x': 3, 'y': 3, 'scalex': 1.5, 'scaley':1.5, 'fraudulent':True}
p2= {'x': -3, 'y': -3, 'scalex': 1.5, 'scaley':1.5, 'fraudulent':True}
p3= {'x': -5, 'y': 5, 'scalex': 2.5, 'scaley':2.5, 'fraudulent':False}
p4= {'x': 5, 'y': -5, 'scalex': 2.5, 'scaley':2.5, 'fraudulent':False}
points = (p1, p2,p3,p4)


data=pd.DataFrame(generator.generateData(points=points, imbalanced=False))
```

```python
import matplotlib.pyplot as plt

# giving a larger plot
plt.figure(figsize=(8, 6))

scatter=plt.scatter(data["x"], data["y"],
            c=data["Fraudulent"],
            cmap='cool')
      #      cmap='plasma')

# labeling x and y axes
plt.xlabel('x')
plt.ylabel('y')
classes = ['Geniune', 'Fraudulent']
plt.legend(handles=scatter.legend_elements()[0], labels=classes)
plt.show()
```

```python
from sklearn.tree import DecisionTreeClassifier


modelClf = DecisionTreeClassifier(random_state=42, max_depth=3)
modelClf.fit(data[["x","y"]], data["Fraudulent"])
```

```python
import matplotlib.pyplot as plt
from sklearn.tree import plot_tree
# Plot the decision tree
plt.figure(figsize=(20, 10))
plot_tree(modelClf, filled=True, feature_names=data.columns, class_names=['Geniune', 'Fraudulent'])
plt.show()
```
