```python
import numpy as np
import pandas as pd

def generateX(fraudulent):
    if(fraudulent):
        return np.random.normal(loc=3, scale=4.0, size=None)
    else:
        return np.random.normal(loc=-3, scale=4.0, size=None)

def generateY(fraudulent):
    if(fraudulent):
        return np.random.normal(loc=3.0, scale=4.5, size=None)
    else:
        return np.random.normal(loc=-3.0, scale=4.5, size=None)

mylist = []
for i in range(0, 10_000, 1):
    if(i%100==0):
        data={"x":generateX(True), "y":generateY(True),"Fraudulent":True}
        mylist.append(data)
    else:
        data={"x":generateX(False), "y":generateY(False),"Fraudulent":False}
        mylist.append(data)


df=pd.DataFrame(mylist)
df.head()
```

```python
import matplotlib.pyplot as plt

# giving a larger plot
plt.figure(figsize=(8, 6))

scatter=plt.scatter(df["x"], df["y"],
            c=df["Fraudulent"],
            cmap='cool')
      #      cmap='plasma')

# labeling x and y axes
plt.xlabel('x')
plt.ylabel('y')
classes = ['Geniune', 'Fraudulent']
plt.legend(handles=scatter.legend_elements()[0], labels=classes)
plt.show()
```

```python
from imblearn.under_sampling import RandomUnderSampler
from datetime import datetime

predictors = ["x","y"]
target = "Fraudulent"

x1, y1 = df[predictors], df[target]
undersample = RandomUnderSampler(sampling_strategy=0.5,random_state=42)
x2, y2 = undersample.fit_resample(x1,y1)
```

```python
plt.figure(figsize=(8, 6))
scatter=plt.scatter(x2["x"], x2["y"],
            c=y2,
            cmap='cool')
      #      cmap='plasma')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
classes = ['Geniune', 'Fraudulent']
plt.legend(handles=scatter.legend_elements()[0], labels=classes)
plt.show()
```
