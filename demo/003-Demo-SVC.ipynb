```python
import numpy as np
import pandas as pd

def generateX(fraudulent):
    if(fraudulent):
        return np.random.normal(loc=3, scale=6.0, size=None)
    else:
        return np.random.normal(loc=-3, scale=5.0, size=None)

def generateY(fraudulent):
    if(fraudulent):
        return np.random.normal(loc=3.0, scale=8.0, size=None)
    else:
        return np.random.normal(loc=-3.0, scale=7.0, size=None)

mylist = []
for i in range(0, 100, 1):
    data1={"x":generateX(False), "y":generateY(False),"Fraudulent":False}
    mylist.append(data1)
    data2={"x":generateX(True), "y":generateY(True),"Fraudulent":True}
    mylist.append(data2)

data=pd.DataFrame(mylist)
data.head()
```

```python
import matplotlib.pyplot as plt

# giving a larger plot
plt.figure(figsize=(8, 6))

scatter=plt.scatter(data["x"], data["y"],
            c=data["Fraudulent"],
            cmap='cool')
      #      cmap='plasma')

# labeling x and y axes
plt.xlabel('x')
plt.ylabel('y')
classes = ['Geniune', 'Fraudulent']
plt.legend(handles=scatter.legend_elements()[0], labels=classes)
plt.show()
```

```python
from sklearn.svm import LinearSVC


modelClf = LinearSVC(random_state=42)
modelClf.fit(data[["x","y"]], data["Fraudulent"])
intercept=modelClf.intercept_
coef=modelClf.coef_[0]

xx = np.linspace(-10,10)
yy = -coef[[0]]/coef[[1]] * xx - intercept/coef[1]
```

```python
plt.figure(figsize=(8, 6))
scatter=plt.scatter(data["x"], data["y"],
            c=data["Fraudulent"],
            cmap='cool')
      #      cmap='plasma')
plt.plot(xx, yy, color='red', label='Line of Best Fit')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.title("Example Linear SVC")
classes = ['Geniune', 'Fraudulent']
plt.legend(handles=scatter.legend_elements()[0], labels=classes)
plt.show()
```
