# Simple example balanced data

```python
import numpy as np
import pandas as pd

p1= {'x': 5, 'y': 5, 'scalex': 2.5, 'scaley':3.0, 'fraudulent':False}
p2= {'x': -5, 'y': -5, 'scalex': 2.5, 'scaley':3.0, 'fraudulent':True}
points = (p1, p2)


data=pd.DataFrame(generator.generateData(points=points, imbalanced=False))
data.head()
import matplotlib.pyplot as plt

# giving a larger plot
plt.figure(figsize=(8, 6))

scatter=plt.scatter(data["x"], data["y"],
            c=data["Fraudulent"],
            cmap='cool')
      #      cmap='plasma')

# labeling x and y axes
plt.xlabel('x')
plt.ylabel('y')
classes = ['Geniune', 'Fraudulent']
plt.legend(handles=scatter.legend_elements()[0], labels=classes)
plt.show()
```

## sklearn.datasets make_moons example

```python
from sklearn.datasets import make_moons
X, y = make_moons(n_samples=200, noise=0.1, random_state=42)

data=pd.DataFrame(dict(x=X[:,0], y=X[:,1],Fraudulent=np.bool(y))) 


import matplotlib.pyplot as plt

# giving a larger plot
plt.figure(figsize=(8, 6))

scatter=plt.scatter(data["x"], data["y"],
            c=data["Fraudulent"],
            cmap='cool')
      #      cmap='plasma')

# labeling x and y axes
plt.xlabel('x')
plt.ylabel('y')
classes = ['Geniune', 'Fraudulent']
plt.legend(handles=scatter.legend_elements()[0], labels=classes)
plt.show()
```
