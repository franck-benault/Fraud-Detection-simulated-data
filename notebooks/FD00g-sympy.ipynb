# Simple formula

```python
from sympy import symbols, log, E

x = symbols('x')
mylogx=log(x)

display(mylogx)

fvars = symbols('x y')
valuex = [10.0,1.0]
print(mylogx,'=',mylogx.evalf(subs = dict(zip(fvars,valuex))))

valuex = [E.evalf(),1.0]
print(mylogx,'=',mylogx.evalf(subs = dict(zip(fvars,valuex))))
```

# Entropy

```python
from sympy import symbols, expand, factor,Sum, log
from sympy import init_printing
init_printing() 

i,n, p = symbols('i n p')
exp1 = symbols('p_i')
#display(exp1)

exp2= log(exp1,2)
#display(exp2)
entropy = -1*Sum(exp1*exp2, (i, 1,n))
display(entropy)
```

```python
from sympy import symbols, expand, factor,Sum, log
from sympy import init_printing
init_printing() 


exp1 = symbols('p_1')
#display(exp1)
exp2 = symbols('p_2')

entropy = -exp1*log(exp1,2)-exp2*log(exp2,2)
display(entropy)

fvars = symbols('p_1, p_2')
b = [0.5,0.5]
print(entropy.evalf(subs = dict(zip(fvars,b))))

b = [0.1,0.9]
print(entropy.evalf(subs = dict(zip(fvars,b))))

b = [0.01,0.99]
print(entropy.evalf(subs = dict(zip(fvars,b))))

b = [0.001,0.999]
print(entropy.evalf(subs = dict(zip(fvars,b))))

       
```

# Gini

```python
from sympy import symbols, expand, factor,Sum, log
from sympy import init_printing
init_printing() 


exp1 = symbols('p_1')
#display(exp1)
exp2 = symbols('p_2')

gini= 1-exp1**2-exp2**2
display(gini)
b = [0.5,0.5]
print(gini.evalf(subs = dict(zip(fvars,b))))
b = [0.1,0.9]
print(gini.evalf(subs = dict(zip(fvars,b))))
b = [0.01,0.99]
print(gini.evalf(subs = dict(zip(fvars,b))))
b = [0.001,0.999]
print(gini.evalf(subs = dict(zip(fvars,b))))
```
