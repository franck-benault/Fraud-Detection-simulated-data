__[Openml example 45955](https://www.openml.org/search?type=data&status=active&id=45955)__



```python
import openml

dataset = openml.datasets.get_dataset(45955)


# Get dataset by name
#dataset = openml.datasets.get_dataset('Fashion-MNIST')
# Get the data itself as a dataframe (or otherwise)
df, _, _, _ = dataset.get_data(dataset_format="dataframe")
print(df.shape)
df.head()
```

# EDA (Exploratory Data Analysis)

```python
import matplotlib.pyplot as plt

print("Pie chart fraud are very unbalanced data:")
print(df['fraud'].value_counts())
plt.pie(df['fraud'].value_counts(),autopct='%1.1f%%', labels=['Genuine','Fraud'], colors=['yellowgreen','r'])
plt.title('Openml dataset 45955')
plt.show()
```

```python
df.columns
```

```python
column_nan_count = df.isnull().sum()
print("NaN count per column:")
print(column_nan_count)
# nothing found
```

```python
import seaborn as sns
import matplotlib.pyplot as plt


matrix = df.corr()

plt.figure(figsize=(8,6))
sns.heatmap(matrix, annot=True, cmap="coolwarm", fmt=".2f", linewidths=0.5)
plt.title("Correlation Heatmap")
plt.show()
```
