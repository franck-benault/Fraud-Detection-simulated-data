## Functions

### Exponent and logarithm

```python
import numpy as np

print(np.exp(np.log(1)))
print(np.exp(np.log(10)))
```

```python
import numpy as np

print(np.log(np.e))
print(np.log(np.e*np.e))
```

### Square root

```python
import numpy as np

print(np.sqrt(4))
print(np.sqrt(16))
```
