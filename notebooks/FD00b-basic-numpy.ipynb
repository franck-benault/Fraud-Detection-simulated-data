## Functions

### Exponent and logarithm

```python
import numpy as np

print(np.exp(np.log(1)))
print(np.exp(np.log(10)))
```

```python
import numpy as np

print(np.log(np.e))
print(np.log(np.power(np.e,3)))
```

### Square root

```python
import numpy as np

print(np.sqrt(4))
print(np.sqrt(16))
```

## Power

```python
print(np.power(4,2))
```

### Numpy not perfect calculation

```python
import numpy as np

print(np.power(np.sqrt(3),2))
print(np.exp(np.log(3)))
```
