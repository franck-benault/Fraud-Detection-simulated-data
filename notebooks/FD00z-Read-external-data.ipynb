## 45955

__[Openml example 45955](https://www.openml.org/search?type=data&status=active&id=45955)__

```python
import openml

dataset = openml.datasets.get_dataset(45955)


# Get dataset by name
#dataset = openml.datasets.get_dataset('Fashion-MNIST')
# Get the data itself as a dataframe (or otherwise)
df, _, _, _ = dataset.get_data(dataset_format="dataframe")
print(df.shape)
print(df['fraud'].value_counts())
print(df['fraud'].value_counts(normalize=True))
df.head()
```

## 46455

__[Openml example 464555](https://www.openml.org/search?type=data&status=active&id=46455)__



```python
import openml

dataset = openml.datasets.get_dataset(46455)


# Get dataset by name
#dataset = openml.datasets.get_dataset('Fashion-MNIST')
# Get the data itself as a dataframe (or otherwise)
df, _, _, _ = dataset.get_data(dataset_format="dataframe")
print(df.shape)
print(df['class'].value_counts())
print(df['class'].value_counts(normalize=True))
df.head()
```

## Split data

link https://scikit-learn.org/stable/modules/generated/sklearn.model_selection.train_test_split.html

```python
from sklearn.model_selection import train_test_split

TEST_SIZE=0.2
target='class'
RANDOM_STATE=42

x_train, x_test, y_train, y_test = train_test_split(df, df[target], test_size = TEST_SIZE, 
                                                        random_state = RANDOM_STATE)

print(df['class'].value_counts(normalize=True))
print(x_train['class'].value_counts(normalize=True))
print(x_test['class'].value_counts(normalize=True))
```

```python
from sklearn.model_selection import train_test_split

TEST_SIZE=0.2
target='class'
RANDOM_STATE=42

x_train, x_test, y_train, y_test = train_test_split(df, df[target], test_size = TEST_SIZE, 
                                                        stratify=df[target],
                                                        random_state = RANDOM_STATE)

print(df['class'].value_counts(normalize=True))
print(x_train['class'].value_counts(normalize=True))
print(x_test['class'].value_counts(normalize=True))
```
