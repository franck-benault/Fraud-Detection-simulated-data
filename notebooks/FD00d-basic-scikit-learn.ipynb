## Data preprocessing

### Categorical feature encoding

```python
import numpy as np
import pandas as pd

data = { 'Amount': np.array([1.0, 2.2, 4.4, 6.7]),
         'Country': np.array(['FR', 'BE', 'BE', 'UK']) }
df = pd.DataFrame(data)
print(df)
```

```python
from sklearn.preprocessing import OneHotEncoder


categorical_columns = df.select_dtypes(include=['object']).columns
print(categorical_columns)

encoder = OneHotEncoder(sparse_output=True, handle_unknown='error')
data = encoder.fit_transform(df[categorical_columns])
print(data.toarray())

dfEncoded = pd.DataFrame(
    data.toarray(),
    columns=encoder.get_feature_names_out(categorical_columns)
)
print(dfEncoded)

dfFinal = pd.concat(
    [df.drop(columns=categorical_columns), dfEncoded],
    axis=1
)
print(dfFinal)
```

## Scaling

```python
import numpy as np
import pandas as pd

data = { 'Amount': np.array([1.0, 2.5, 7.0, 10.0]),
         'Distance': np.array([101.0, 252, 701, 1001]),
         }
df = pd.DataFrame(data)
print(df)
```

```python
from sklearn.preprocessing import MinMaxScaler

scaler=MinMaxScaler(feature_range=(-1,1))
dfWork = scaler.fit_transform(df)
print(dfWork)
```

```python
from sklearn.preprocessing import StandardScaler

scaler=StandardScaler()
dfWork = scaler.fit_transform(df)
print(dfWork)
```

## Transformer

```python
import numpy as np
import pandas as pd
from sklearn.preprocessing import FunctionTransformer

data = { 'Amount': np.array([4.0, 9.0, 16.0, 25.0]),
         'Distance': np.array([4.0, 16.0, 4.0, 36.0]),
         }
df = pd.DataFrame(data)
print('Before transformation')
print(df)

transformer= FunctionTransformer(np.sqrt)
df = transformer.transform(df)

print('After transformation')
print(df)
```

```python
import numpy as np
import pandas as pd
from sklearn.preprocessing import FunctionTransformer

data = { 'Amount': np.array([np.e, np.e*np.e, np.e, np.e]),
         'Distance': np.array([np.e, np.e*np.e*np.e, 1, np.e]),
         }
df = pd.DataFrame(data)
print('Before transformation')
print(df)

transformer= FunctionTransformer(np.log)
df = transformer.transform(df)

print('After transformation')
print(df)
```
