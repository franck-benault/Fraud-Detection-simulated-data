```python
from sklearn.metrics import f1_score
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import train_test_split

dfWork=df.copy()


TEST_SIZE = 0.20 # test size using_train_test_split
RANDOM_STATE = 0

predictors=['distance_from_home', 'distance_from_last_transaction',
       'ratio_to_median_purchase_price', 'repeat_retailer', 'used_chip',
       'used_pin_number', 'online_order']
target='fraud'

x_train, x_test, y_train, y_test = train_test_split(dfWork[predictors], dfWork[target], test_size = TEST_SIZE, 
                                                        stratify=dfWork[target],
                                                        random_state = RANDOM_STATE)

hyperparametersFound = {'max_depth': 4}
modelClf = DecisionTreeClassifier(random_state=42)
parameters=hyperparametersFound
modelClf.set_params(**parameters)

modelClf.fit(x_train, y_train)

predsTrain = modelClf.predict(x_train)
predsTest = modelClf.predict(x_test)


f1Learning =f1_score(y_train, predsTrain)
f1Test=f1_score(y_test, predsTest)

print(f"f1 learning {f1Learning:.3f}")
print(f"f1 test {f1Test:.3f}")
dm.show_confusion_matrix(y_train, predsTrain,'Confusion matrix learning data')
dm.show_confusion_matrix(y_test, predsTest,'Confusion matrix test data')
```

```python
from sklearn.metrics import f1_score
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import train_test_split

TEST_SIZE = 0.20 # test size using_train_test_split
RANDOM_STATE = 0

dfWork=df.copy()
dfWork['log_distance_from_home'] = np.log(dfWork['distance_from_home'])
dfWork['log_distance_from_last_transaction'] = np.log(dfWork['distance_from_last_transaction'])
dfWork['log_ratio_to_median_purchase_price'] = np.log(dfWork['ratio_to_median_purchase_price'])

predictors=['log_distance_from_home', 'log_distance_from_last_transaction',
       'log_ratio_to_median_purchase_price', 'used_chip',
       'used_pin_number', 'online_order']
target='fraud'

x_train, x_test, y_train, y_test = train_test_split(dfWork[predictors], dfWork[target], test_size = TEST_SIZE, 
                                                        stratify=dfWork[target],
                                                        random_state = RANDOM_STATE)


hyperparametersFound = {'max_depth': 4}
modelClf = DecisionTreeClassifier(random_state=42)
parameters=hyperparametersFound
modelClf.set_params(**parameters)

modelClf.fit(x_train, y_train)

predsTrain = modelClf.predict(x_train)
predsTest = modelClf.predict(x_test)


f1Learning =f1_score(y_train, predsTrain)
f1Test=f1_score(y_test, predsTest)

print(f"f1 learning {f1Learning:.3f}")
print(f"f1 test {f1Test:.3f}")
dm.show_confusion_matrix(y_train, predsTrain,'Confusion matrix learning data')
dm.show_confusion_matrix(y_test, predsTest,'Confusion matrix test data')
```
