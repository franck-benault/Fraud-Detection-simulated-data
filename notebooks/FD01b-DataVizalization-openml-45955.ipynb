# Data correlation

## Correlation matrix

```python
# Clean feature remove the categorical values

dfTemp=df.drop(['repeat_retailer','used_chip','used_pin_number','online_order','fraud'],
  axis='columns', inplace=False)

corr_matrix=dfTemp.corr()
corr_matrix['distance_from_home'].sort_values(ascending=False)
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

dfTemp=df.drop(['repeat_retailer','used_chip','used_pin_number','online_order','fraud'],
  axis='columns', inplace=False)

corr_matrix=dfTemp.corr()

plt.figure(figsize=(8,6))
sns.heatmap(corr_matrix, annot=True, cmap="coolwarm", fmt=".3f", linewidths=0.5)
plt.title("Correlation Heatmap")
plt.show()
```

```python
from sklearn.preprocessing import FunctionTransformer
import numpy as np

dftemp=df.drop(['repeat_retailer','used_chip','used_pin_number','online_order','fraud'],
  axis='columns', inplace=False)

log_transformer= FunctionTransformer(np.log, inverse_func=np.exp)
dftemp_transformed = log_transformer.transform(dftemp)

columns = dftemp_transformed.columns
for column in columns:
    dftemp_transformed[column].plot.density(color='green')
    plt.show()
```

```python

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from matplotlib import gridspec

import warnings
warnings.filterwarnings('ignore')


data_plot=df.drop(['repeat_retailer', 'used_chip','used_pin_number', 'online_order'],  axis='columns', inplace=False)


data_plot['log_distance_from_home'] = np.log(data_plot['distance_from_home'])
data_plot['log_distance_from_last_transaction'] = np.log(data_plot['distance_from_last_transaction'])
data_plot['log_ratio_to_median_purchase_price'] = np.log(data_plot['ratio_to_median_purchase_price'])


# Plot the distributions of the features
columns = data_plot.iloc[:,0:30].columns
plt.figure(figsize=(12,30*4))
grids = gridspec.GridSpec(30, 1)
for grid, index in enumerate(data_plot[columns]):
 ax = plt.subplot(grids[grid])
 sns.distplot(data_plot[index][data_plot.fraud == 1], hist=False, kde_kws={"shade": True}, bins=50)
 sns.distplot(data_plot[index][data_plot.fraud == 0], hist=False, kde_kws={"shade": True}, bins=50)
 ax.set_xlabel("")
 ax.set_title("Distribution of Column: "  + str(index))
plt.show()
```
