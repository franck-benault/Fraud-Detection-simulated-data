# Series

## Example Series creation

```python
import pandas as pd

data_list = ['g', 'e', 'e', 'k', 's','r', 'r', 'e']
ser = pd.Series(data_list)
print(ser)
```

```python
# Copy() Duplicate the Serie

data_list = ['g', 'e', 'e', 'k', 's','r', 'r', 'e']
ser = pd.Series(data_list)
ser2=ser.copy()
print(ser2)
```

## pandas.Series.value_counts
Series.value_counts(normalize=False, sort=True, ascending=False, bins=None, dropna=True)

```python
import pandas as pd

data_list = ['g', 'e', 'e', 'k', 's','r', 'r', 'e']
ser = pd.Series(data_list)
ser.value_counts(normalize=True)
```

# DataFrame

## Example dataFrame creation

```python
import numpy as np
import pandas as pd

data = { 'A': np.array([1, 2, 4]),
         'B': np.array([2, 3, 7]),
         'C': np.array(['A', 'E', 'A']) }
df = pd.DataFrame(data)
print(df)
```

```python
# Copy() Duplicate the Serie

import numpy as np
import pandas as pd

data = { 'A': np.array([1, 2, 4]),
         'B': np.array([2, 3, 7]),
         'C': np.array(['A', 'E', 'A']) }
df = pd.DataFrame(data)
df2=df.copy()
print(df2)
```
